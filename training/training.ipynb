```python
import pandas as pd

doc= pd.read_csv('dataset\synthetic_logs.csv')
doc.head(3)
```

```python
from sklearn.cluster import DBSCAN
from sentence_transformers import SentenceTransformer
```

```python
model = SentenceTransformer('all-MiniLM-L6-v2')  # Lightweight embedding models
embeddings = model.encode(doc['log_message'].tolist())
```

```python
embeddings[:3]
```

```python
clustering = DBSCAN(eps=0.2, min_samples=1, metric='cosine').fit(embeddings)
doc['cluster'] = clustering.labels_
```

```python
doc.head()
```

```python
# Group by cluster to inspect patterns
clusters = doc.groupby('cluster')['log_message'].apply(list)
sorted_clusters = clusters.sort_values(key=lambda x: x.map(len), ascending=False)
```

```python
print("Clustered Patterns:")
for cluster_id, messages in sorted_clusters.items():
    if len(messages) > 10:
        print(f"Cluster {cluster_id}:")
        for msg in messages[:5]:
            print(f"  {msg}")
```

```python
import re
def classify_with_regex(log_message):
    regex_patterns = {
        r"User User\d+ logged (in|out).": "User Action",
        r"Backup (started|ended) at .*": "System Notification",
        r"Backup completed successfully.": "System Notification",
        r"System updated to version .*": "System Notification",
        r"File .* uploaded successfully by user .*": "System Notification",
        r"Disk cleanup completed successfully.": "System Notification",
        r"System reboot initiated by user .*": "System Notification",
        r"Account with ID .* created by .*": "User Action"
    }
    for pattern, label in regex_patterns.items():
        if re.search(pattern, log_message, re.IGNORECASE):
            return label
    return None
```

```python
print(classify_with_regex("File data_6169.csv uploaded successfully by user User953."))
```

```python
doc['regex_label']=doc['log_message'].apply(classify_with_regex)
doc[doc.regex_label.notnull()]
```

```python
doc_not_regex=doc[doc['regex_label'].isnull()].copy()
doc_not_regex
```

```python

df_legacy = doc_not_regex[doc_not_regex.source=="LegacyCRM"]
df_legacy
```

```python
df_non_legacy = doc_not_regex[doc_not_regex.source!="LegacyCRM"]
df_non_legacy
```

```python

df_non_legacy.shape
```

```python
model = SentenceTransformer('all-MiniLM-L6-v2')  # Lightweight embedding models
embeddings_filtered = model.encode(df_non_legacy['log_message'].tolist())
```

```python
len(embeddings_filtered)
```

```python

X = embeddings_filtered
y = df_non_legacy['target_label'].values
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
clf=LogisticRegression(max_iter=1000)
clf.fit(X_train,y_train)
y_pred = clf.predict(X_test)
report = classification_report(y_test, y_pred)
print(report)
```

```python
import joblib
joblib.dump(clf, '../training/log_classifier.joblib')
```
